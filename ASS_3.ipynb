```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('Student_Performance_Dataset.csv')
print(df.head(10))
```

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns = df.columns.str.lower()
```

```python
mean_val = df['math_score'].mean()
print("Mean of math score:", mean_val)
mode_val = df['math_score'].mode()[0]
print("Mode of math score:", mode_val)  
median_val = df['math_score'].median()
print("Median of math score:", median_val)
```

```python
print(df.columns.tolist())
```

```python
[col for col in df.columns if 'gen' in col.lower() or 'sex' in col.lower()]
```

```python
group_summary = df.groupby('gender')['math_score'].agg([
    'mean',
    'median',
    lambda x: x.mode()[0]
])
```

```python
print(group_summary)    
```

```python
group_summary.columns = ['Mean', 'Median', 'Mode']
print(group_summary)
```

```python
print("\nGroup-wise Analysis:")
print(group_summary)
```

```python
skewness = df['math_score'].skew()
print('\nSkewness:', skewness)

if skewness > 0:
    print('Distribution is right (positively) skewed')
elif skewness < 0:
    print('Distribution is left (negatively) skewed')
else:
    print('Distribution is approximately symmetric')
```
