```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Titanic-Dataset.csv')

print(df.head())
print(df.columns)
```

```python
survival_rate = df.groupby('Pclass')['Survived'].mean() * 100

plt.figure(figsize=(8, 5))
plt.plot(survival_rate.index, survival_rate.values,
         marker='o', color='teal')

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')
plt.xticks([1, 2, 3])
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='Pclass', y='Fare',
            hue='Pclass', palette='viridis', legend=False)

plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['Age'].dropna(), bins=10,
             kde=True, color='orange')

plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='Age', y='Fare',
                hue='Survived', palette='Set1',
                alpha=0.7)

plt.title('Passenger Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Pclass', y='Fare',
            hue='Pclass', palette='Set2', legend=False)

plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

```python
survival_counts = df['Survived'].value_counts().sort_index()

plt.figure(figsize=(6, 6))

plt.pie(
    survival_counts,
    labels=['Did Not Survive', 'Survived'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Passenger Survival Distribution')
plt.show()
```

```python
numeric_data = df[
    ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
]

plt.figure(figsize=(8, 6))

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Titanic Dataset')
plt.show()
```
